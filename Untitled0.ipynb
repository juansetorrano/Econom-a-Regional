```python
def consultar_indicador(clave, codigo_estado):
    base = (
        "https://www.inegi.org.mx/app/api/indicadores/"
        "desarrolladores/jsonxml/INDICATOR/"
    )

    url = f"{base}{clave}/es/{codigo_estado}/true/BISE/2.0/{token}"
    respuesta = requests.get(url, params={"type": "json"}, timeout=30)

    if respuesta.status_code != 200:
        raise ValueError(f"Falló la consulta del estado {codigo_estado}")

    datos = respuesta.json()

    if not datos.get("Series"):
        raise ValueError(f"No hay serie para {clave}, estado {codigo_estado}")

    observaciones = datos["Series"][0]["OBSERVATIONS"]

    if len(observaciones) != 1:
        raise ValueError(f"Revisar observaciones del estado {codigo_estado}")

    dato = observaciones[0]

    return {
        "CVE_ENT": codigo_estado,
        "PERIODO": dato["TIME_PERIOD"],
        "VALOR": float(dato["OBS_VALUE"])
    }
```

```python
descargas = {}
errores = {}
```

```python
for nombre, clave in indicadores.items():
    for codigo in estados["CVE_ENT"]:
        pareja = (nombre, codigo)

        # Evita repetir consultas que ya funcionaron
        if pareja in descargas:
            continue

        try:
            dato = consultar_indicador(clave, codigo)
            dato["INDICADOR"] = nombre
            descargas[pareja] = dato
            errores.pop(pareja, None)

        except Exception as error:
            errores[pareja] = type(error).__name__

    cantidad = sum(
        (nombre, codigo) in descargas
        for codigo in estados["CVE_ENT"]
    )
    print(f"{nombre}: {cantidad}/32 estados descargados")

datos_bienestar = pd.DataFrame(list(descargas.values()))

print("Datos obtenidos:", len(datos_bienestar))
print("Consultas con error:", len(errores))
```

```python
valores = datos_bienestar.pivot(
    index="CVE_ENT",
    columns="INDICADOR",
    values="VALOR"
).reindex(columns=list(indicadores))

bienestar_final = estados[["CVE_ENT", "NOM_ENT"]].merge(
    valores,
    on="CVE_ENT",
    how="left",
    validate="one_to_one"
)
```

```python
print(bienestar_final.shape)
print(bienestar_final.isna().sum().sum())
bienestar_final.head()
```

```python
periodos = datos_bienestar.pivot(
    index="CVE_ENT",
    columns="INDICADOR",
    values="PERIODO"
).reindex(columns=list(indicadores))
```

```python
correcciones = {
    "09": "Ciudad de México",
    "15": "México",
    "16": "Michoacán de Ocampo",
    "22": "Querétaro",
    "24": "San Luis Potosí",
    "19": "Nuevo León",
    "31": "Yucatán"
}

for codigo, nombre in correcciones.items():
    bienestar_final.loc[
        bienestar_final["CVE_ENT"] == codigo, "NOM_ENT"
    ] = nombre

with pd.ExcelWriter("Bienestar.xlsx") as archivo:
    bienestar_final.to_excel(archivo, sheet_name="Bienestar", index=False)
    periodos.to_excel(archivo, sheet_name="Periodos")

from google.colab import files
files.download("Bienestar.xlsx")
```
